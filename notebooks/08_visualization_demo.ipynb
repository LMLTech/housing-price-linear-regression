**Đặc biệt:** Báo cáo này đã được chia tách rõ ràng thành hai phần để đánh giá độ chính xác của mô hình dự báo tại **TP.HCM** và **Hà Nội** riêng biệt.


```python
import matplotlib.pyplot as plt
from IPython.display import Image, display
from pathlib import Path

FIGURES_DIR = Path('../reports/figures')

def show_figure(fig_name):
    fig_path = FIGURES_DIR / fig_name
    if fig_path.exists():
        display(Image(filename=str(fig_path)))
    else:
        print(f"[CẢNH BÁO] Không tìm thấy hình ảnh: {fig_name}")
```

## PHẦN 1: PHÂN TÍCH KHÁM PHÁ DỮ LIỆU (EDA - TOÀN QUỐC)
Các biểu đồ dưới đây mô tả dữ liệu gốc trước khi chia tách mô hình, giúp chúng ta hiểu rõ bức tranh toàn cảnh về thị trường bất động sản.

```python
print("--- 1. Phân phối Giá nhà ---")
show_figure("01_price_distribution.png")
print("--- 2. Phân phối Diện tích (Sau khi loại bỏ Outlier) ---")
show_figure("02_area_distribution.png")
print("--- 3. Tương quan Giá - Diện tích ---")
show_figure("03_price_vs_area.png")
print("--- 4. Bản đồ Nhiệt Tương quan (Correlation Heatmap) ---")
show_figure("04_correlation_heatmap.png")
print("--- 5. So sánh Giá theo Tỉnh/Thành ---")
show_figure("05_price_by_province.png")
print("--- 6. So sánh Giá theo Loại Bất động sản ---")
show_figure("06_price_by_property_type.png")
print("--- 7. Giá theo Số phòng ngủ ---")
show_figure("07_price_by_bedrooms.png")
print("--- 8. Tương quan Khoảng cách tới Trung tâm & Giá ---")
show_figure("08_distance_vs_price.png")
print("--- 9. Số lượng Mẫu dữ liệu theo Quận/Huyện ---")
show_figure("09_count_by_district.png")
```

```python
print("--- 10. Giá Thực tế vs. Giá Dự đoán (TP.HCM) ---")
show_figure("10_actual_vs_predicted_hcm.png")
print("--- 11. Phân phối Sai số (TP.HCM) ---")
show_figure("11_residual_distribution_hcm.png")
print("--- 12. Sai số vs. Giá Dự đoán (TP.HCM) ---")
show_figure("12_residual_vs_predicted_hcm.png")
print("--- 13. Phân phối Sai số Tuyệt đối (TP.HCM) ---")
show_figure("13_absolute_error_distribution_hcm.png")
print("--- 14. Sai số theo Phân khúc Giá (TP.HCM) ---")
show_figure("14_error_by_price_range_hcm.png")
print("--- 15. Top 20 Căn nhà Dự đoán Sai nhiều nhất (TP.HCM) ---")
show_figure("15_top_prediction_errors_hcm.png")
```
